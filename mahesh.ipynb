```python
import numpy as np
import pandas as pd


number_of_rolls = 1000
rng = np.random.default_rng()

# Generate integers from 1 through 6, inclusive.
rolls = rng.integers(1, 7, size=number_of_rolls)

results = pd.DataFrame({"Roll": rolls})
frequency_table = (
	results["Roll"]
	.value_counts()
	.reindex(range(1, 7), fill_value=0)
	.rename_axis("Face")
	.reset_index(name="Frequency")
)
frequency_table["Relative Frequency"] = (
	frequency_table["Frequency"] / number_of_rolls
)

print("First 10 rolls:")
print(results.head(10).to_string(index=False))
print("\nDice frequency table:")
print(frequency_table.to_string(index=False))
```

```python
import numpy as np

# Sample space for 3 coin tosses
sample_space = np.array([
    "HHH", "HHT", "HTH", "HTT",
    "THH", "THT", "TTH", "TTT"
])

print("Sample Space:")
print(sample_space)

# Total outcomes
total = len(sample_space)

# a) Exactly 3 Heads
exactly_3 = np.array([x.count("H") == 3 for x in sample_space])
p3 = np.sum(exactly_3) / total

# b) Exactly 2 Heads
exactly_2 = np.array([x.count("H") == 2 for x in sample_space])
p2 = np.sum(exactly_2) / total

# c) At least 1 Head
at_least_1 = np.array([x.count("H") >= 1 for x in sample_space])
p1 = np.sum(at_least_1) / total

print("Probability of exactly 3 Heads:", p3)
print("Probability of exactly 2 Heads:", p2)
print("Probability of at least 1 Head:", p1)
```

```python
import numpy as np

# Sample space of a die
die = np.array([1, 2, 3, 4, 5, 6])

total = len(die)

# a) Even number
even = np.sum(die % 2 == 0)
print("Probability of even number:", even / total)

# b) Odd number
odd = np.sum(die % 2 != 0)
print("Probability of odd number:", odd / total)

# c) Number greater than 4
greater_4 = np.sum(die > 4)
print("Probability of number greater than 4:", greater_4 / total)

# d) Number less than or equal to 3
less_equal_3 = np.sum(die <= 3)
print("Probability of number <= 3:", less_equal_3 / total)
```

```python
import numpy as np

# Sample space of two dice
dice1 = np.array([1, 2, 3, 4, 5, 6])
dice2 = np.array([1, 2, 3, 4, 5, 6])

sample_space = np.array(np.meshgrid(dice1, dice2)).T.reshape(-1, 2)

print("Sample Space:")
print(sample_space)

total = len(sample_space)

# a) Sum is 7
a = np.sum(sample_space[:, 0] + sample_space[:, 1] == 7)
print("a) Probability of sum 7:", a / total)

# b) Sum is greater than 8
b = np.sum(sample_space[:, 0] + sample_space[:, 1] > 8)
print("b) Probability of sum greater than 8:", b / total)

# c) Both dice show same number
c = np.sum(sample_space[:, 0] == sample_space[:, 1])
print("c) Probability of same number:", c / total)

# d) At least one die shows 6
d = np.sum((sample_space[:, 0] == 6) | (sample_space[:, 1] == 6))
print("d) Probability of at least one 6:", d / total)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 10,000 coin tosses
tosses = np.random.randint(0, 2, 10000)

# Cumulative probability of Heads
heads = np.cumsum(tosses)
toss_number = np.arange(1, 10001)
probability = heads / toss_number

# Plot
plt.plot(toss_number, probability)
plt.axhline(0.5, linestyle='--')

plt.xscale('log')
plt.xlabel("Number of Tosses")
plt.ylabel("Cumulative Probability of Heads")
plt.title("Law of Large Numbers")
plt.show()
```

```python
import numpy as np

# Sample space
die = np.array([1, 2, 3, 4, 5, 6])

# A = even number
A = die % 2 == 0

# B = number greater than 3
B = die > 3

# A and B
A_and_B = A & B

# P(A|B) = P(A and B) / P(B)
P_A_given_B = np.sum(A_and_B) / np.sum(B)

print("P(A|B) =", P_A_given_B)
```

```python
import numpy as np

# Given values
P_D = 0.02          # Probability of disease
P_positive_D = 0.95 # Sensitivity
P_positive_not_D = 0.04  # False-positive rate

# Bayes' Theorem
P_positive = (P_positive_D * P_D) + (P_positive_not_D * (1 - P_D))

P_D_positive = (P_positive_D * P_D) / P_positive

print("P(Disease | Positive) =", P_D_positive * 100, "%")
```

```python
import numpy as np

# Given values
P_spam = 0.20
P_offer_spam = 0.80
P_offer_not_spam = 0.10

# Probability of getting "offer"
P_offer = (P_offer_spam * P_spam) + (P_offer_not_spam * (1 - P_spam))

# Bayes theorem
P_spam_offer = (P_offer_spam * P_spam) / P_offer

print("P(Spam | offer) =", P_spam_offer * 100, "%")
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters
n = 20       # Number of trials
p = 0.05      # Probability of success

# Values of x (number of successes)
x = np.arange(0, n + 1)

# Create binomial distribution
pmf = binom.pmf(x, n, p)

# 1. Binomial PMF
plt.figure(figsize=(7, 5))

plt.bar(
    x, pmf,
    color='steelblue',
    edgecolor='black',
    alpha=0.8
)

plt.axvline(
    binom.mean(n, p),
    color='red',
    linestyle='--',
    label=f'Mean = {binom.mean(n, p):.1f}'
)

plt.title(f'Binomial PMF (n={n}, p={p})')
plt.xlabel('Number of Successes (x)')
plt.ylabel('Probability')
plt.xticks(x)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
#from scipy.stats import binom
from scipy import stats

# Visualization - Poisson PMF
plt.figure(figsize=(7, 5))

x_p = np.arange(0, 20)

for lam in [1, 3, 5, 10]:
    plt.plot(x_p, stats.poisson.pmf(x_p, lam),
             marker='o', markersize=4, label=f'λ = {lam}')

plt.title('Poisson PMF — Effect of λ')
plt.xlabel('Number of Events (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import geom

# Probability of success
p_geo = 0.3

# Trial numbers (1 to 19)
x_g = np.arange(1, 20)

# Create Geometric Distribution
geo = geom(p_geo)

# Visualization - Geometric PMF
plt.figure(figsize=(7, 5))

plt.bar(
    x_g, geo.pmf(x_g),
    color='green',
    edgecolor='black',
    alpha=0.8
)

plt.title(f'Geometric PMF (p={p_geo})')
plt.xlabel('Trial Number (x)')
plt.ylabel('Probability')
plt.xticks(x_g)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
